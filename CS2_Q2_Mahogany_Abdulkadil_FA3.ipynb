```python
# Activity 3 (Graded or Non-Graded Formative Assessment): Data Analysis Mini App

# Do your Activity 3 here

scores = [ 95 , 92 , 67 , 90 , 98 ]

def display_scores():
    print("Scores List:", scores)

def search_score():
    item = int(input("Enter score to search: "))
    if item in scores:
        print(f"Found {item} in the list wow")
    else:
        print(f"{item} not found, aw so sad.")

def compute_average():
    slice_range = input("Slicing range to average (start:end): ")
    colon = 0
    for i in range(len(slice_range)):
        if slice_range[i] == ":":
            colon = i

    start = int(slice_range[:colon])
    end = int(slice_range[colon + 1:])
    sliced_scores = scores[start:end]

    average = sum(sliced_scores) / len(sliced_scores)
    print(f"Sliced Scores ({slice_range}):", sliced_scores)
    print("Average Score:", average)

def update_score():
    index = int(input("What index to update? "))
    new_value = int(input("What you want to be new value man? "))
    scores[index] = new_value
    print("Score updated!!")

# Main menu loop to run the program
while True:
    print("\n=== Student Score Analyzer ===")
    print("Select the number of the operation you want to perform:")
    print("1. Display Scores")
    print("2. Search Score")
    print("3. Compute Average Score")
    print("4. Update Score")
    print("5. Display Updated Scores")
    print("6. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        display_scores()
    elif choice == "2":
        search_score()
    elif choice == "3":
        compute_average()
    elif choice == "4":
        update_score()
    elif choice == "5":
        print("Updated Scores List:", scores)
    elif choice == "6":
        print("Exiting program. Goodbye I guess!!")
        break
    else:
        print("Invalid choice, please enter a number from 1 to 6.")
```
